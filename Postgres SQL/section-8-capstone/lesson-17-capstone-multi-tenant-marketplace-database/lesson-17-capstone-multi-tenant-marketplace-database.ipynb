### Setup - Import the Tools
Imports the standard-library helper, the PostgreSQL driver and the `.env` loader.

```python
import os

import psycopg
from dotenv import load_dotenv
```

### Setup - Load the Connection Settings
Reads the global `.env` at the course root. The path is relative, so Jupyter must be running from inside this lesson folder.

```python
# pull the key-value pairs from the global .env (two folders up: lesson folder, then section folder)
load_dotenv("../../.env")

PGHOST = os.getenv("PGHOST", "localhost")
PGPORT = os.getenv("PGPORT", "5432")
PGUSER = os.getenv("PGUSER", "postgres")
PGPASSWORD = os.getenv("PGPASSWORD") or None  # empty means "no password"
PGDATABASE = os.getenv("PGDATABASE", "university")
PGSSLMODE = os.getenv("PGSSLMODE", "prefer")

print(f"Connecting to {PGUSER}@{PGHOST}:{PGPORT}/{PGDATABASE} (sslmode={PGSSLMODE})")
```

### Setup - Connect to PostgreSQL
Opens one connection and asks the server for its version, which proves the settings work. `autocommit=True` makes every statement take effect immediately.

```python
conn = psycopg.connect(
    host=PGHOST,
    port=PGPORT,
    user=PGUSER,
    password=PGPASSWORD,
    dbname=PGDATABASE,
    sslmode=PGSSLMODE,
    autocommit=True,
)

with conn.cursor() as cur:
    cur.execute("SELECT version()")
    print(cur.fetchone()[0])
```
